```python
import torch

def main():
    print("=" * 50)
    print("PyTorch GPU 테스트")
    print("=" * 50)

    # 1. PyTorch 버전 확인
    print(f"PyTorch version: {torch.__version__}")

    # 2. CUDA 버전 확인
    print(f"CUDA version used by PyTorch: {torch.version.cuda}")

    # 3. CUDA 사용 가능 여부
    cuda_available = torch.cuda.is_available()
    print(f"CUDA available: {cuda_available}")

    if not cuda_available:
        print("\nGPU를 사용할 수 없다.")
        print("확인할 것:")
        print("- NVIDIA 드라이버 설치 여부")
        print("- CUDA 지원 PyTorch 설치 여부")
        print("- torch.__version__에 +cu128 같은 표시가 있는지")
        return

    # 4. GPU 정보 출력
    device_count = torch.cuda.device_count()
    print(f"GPU count: {device_count}")

    for i in range(device_count):
        print(f"\nGPU {i}")
        print(f"Name: {torch.cuda.get_device_name(i)}")
        print(f"Capability: {torch.cuda.get_device_capability(i)}")

    # 5. 실제 GPU 연산 테스트
    device = torch.device("cuda")

    print("\n간단한 텐서 연산 테스트 시작")

    a = torch.randn(3000, 3000, device=device)
    b = torch.randn(3000, 3000, device=device)

    torch.cuda.synchronize()
    c = torch.matmul(a, b)
    torch.cuda.synchronize()

    print("행렬 곱셈 완료")
    print(f"Result shape: {c.shape}")
    print(f"Result device: {c.device}")

    # 6. GPU 메모리 사용량 확인
    allocated = torch.cuda.memory_allocated(device) / 1024**2
    reserved = torch.cuda.memory_reserved(device) / 1024**2

    print(f"\nAllocated memory: {allocated:.2f} MB")
    print(f"Reserved memory: {reserved:.2f} MB")

    print("\nGPU 테스트 성공")

if __name__ == "__main__":
    main()
```
